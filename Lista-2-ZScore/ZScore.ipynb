```python
import numpy as np
import pandas as pd
```

```python
# Exercício 1

media = 100
desvio = 5
valor = 115

distancia = valor - media

zscore = distancia / desvio

print(f"Nessa amostragem, o valor {valor} está a {zscore:.2f} desvios padrão da média, o que para a referência de |Z| > 3 ele {'é' if abs(zscore) > 3 else 'não é'} um valor atípico.")
```

```python
# Exercício 2

casos = [85, 100, 120]
media = 100
desvio = 10

for caso in casos:
    distancia = caso - media
    zscore = distancia / desvio
    if zscore < 0:
        print(f"Nessa amostragem, o valor {caso} está a {abs(zscore):.2f} desvios padrão abaixo da média")
    elif zscore > 0:
        print(f"Nessa amostragem, o valor {caso} está a {zscore:.2f} desvios padrão acima da média")
    else:
        print(f"Nessa amostragem, o valor {caso} está exatamente na média")
```

```python
# Exercício 3

temperaturas = [21, 23, 25, 27, 29]
media = 25
desvio = 2

zscores = {}

for temp in temperaturas:
    distancia = temp - media
    zscore = distancia / desvio
    zscores[temp] = zscore

print(f"O valor mais distante da média é: {max(zscores, key=lambda x: abs(zscores[x]))}")
print(f"O maior valor absoluto de Z-score é: {max(zscores.values(), key=abs)}")
```

```python
# Exercício 4

latencias = [98, 102, 101, 99, 100, 103, 97, 180]

media = np.mean(latencias)
desvio = np.std(latencias)

distancia = 180 - media
zscore = distancia / desvio

if abs(zscore) > 3:
    print(f"O valor 180 está a {zscore:.2f} desvios padrão da média, portanto o valor merece ser investigado, porém sem apagar automaticamente.")
else:
    print(f"O valor 180 está a {zscore:.2f} desvios padrão da média, portanto o valor não é considerado atípico.")
```

```python
# Exercício 5

cpu = [42, 45, 47, 44, 46, 43, 48, 92]

media = np.mean(cpu)
desvio = np.std(cpu)

for valor in cpu:
    distancia = valor - media
    zscore = distancia / desvio
    if abs(zscore) > 3:
        print(f"O valor {valor} está a {zscore:.2f} desvios padrão da média, portanto o valor merece ser investigado, porém sem apagar automaticamente.")
    else:
        print(f"O valor {valor} está a {zscore:.2f} desvios padrão da média, portanto o valor não é considerado atípico.")
```

```python
# Exercício 6

media_a = 100
desvio_a = 2

media_b = 100
desvio_b = 20

valor = 110

distancia_a = valor - media_a
distancia_b = valor - media_b

zscore_a = distancia_a / desvio_a
zscore_b = distancia_b / desvio_b

print(f"O valor {valor} está a {zscore_a:.2f} desvios padrão da média do grupo A e a {zscore_b:.2f} desvios padrão da média do grupo B.")
print(f"Mesmo que o valor seja o mesmo, ele é considerado mais atípico no grupo A do que no grupo B, pois o desvio padrão do grupo A é menor, tornando a distância em relação à média mais significativa.")
```

```python
# Exercício 7

def interpretar_z(z):
    if abs(z) > 3:
        return "Investigar"
    if z < 0:
        return "Abaixo da média"
    if z > 0:
        return "Acima da média"
    return "Na média"

valores_z = [-3.5, -1.2, 0, 0.8, 3.7]

for valor in valores_z:
    interpretacao = interpretar_z(valor)
    print(f"O valor Z-score {valor} é classificado como: {interpretacao}")
```

```python
# Exercício 8

dados = {
    "Usuario": ["ana", "bruno", "carla", "diego", "eva", "fabio"],
    "Requisicoes": [120, 135, 128, 122, 130, 400]
}

df = pd.DataFrame(dados)

media_requisicoes = df["Requisicoes"].mean()
desvio_requisicoes = df["Requisicoes"].std()

print(f"Média de requisições: {media_requisicoes:.2f}")
print(f"Desvio padrão de requisições: {desvio_requisicoes:.2f}")

df["Z_Score"] = (df["Requisicoes"] - media_requisicoes) / desvio_requisicoes
df["Status"] = df["Z_Score"].apply(lambda z: "Investigar" if abs(z) > 3 else "Normal")

print(df)

print(df.where(df["Status"] == "Investigar").dropna())
```

```python
# Exercício 9

dados = [10, 11, 12, 12, 13, 13, 14, 15, 30]

Q1, Q3 = np.percentile(dados, [25, 75])
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

media = np.mean(dados)
desvio = np.std(dados)

zscore_30 = (30 - media) / desvio

print(f"Para o valor 30, o Z-score é: {zscore_30:.2f}")
print(f"Para o valor 30, o IQR é: {IQR:.2f}, limite inferior: {limite_inferior:.2f}, limite superior: {limite_superior:.2f}")

print(f"Enquanto o Z-score indica que o valor 30 está dentro de 3 desvios padrão da média, o IQR indica que o valor 30 está fora do intervalo interquartil, sendo considerado um outlier.")
```

```python
# Exercício 10

dados = {
    "Evento": ["A", "B", "C", "D", "E", "F", "G"],
    "Tentativas_Login": [3, 4, 2, 5, 3, 4, 40]
}
df = pd.DataFrame(dados)

media_tentativas = df["Tentativas_Login"].mean()
desvio_tentativas = df["Tentativas_Login"].std()

df["Z_Score"] = (df["Tentativas_Login"] - media_tentativas) / desvio_tentativas
df["Status"] = df["Z_Score"].apply(lambda z: "Investigar" if abs(z) > 3 else "Normal")

print(df)

print("No contexto atual para contagem de tentativas de login, um ZScore alto pode indicar um comportamento suspeito, como uma tentativa de ataque de força bruta, ou apenas um usuário que esqueceu sua senha e tentou várias vezes. Já para outros contextos um ZScore alto é um valor que apenas pode ser um erro na coleta de dados. Por isso é importante analisar o contexto antes de tomar qualquer decisão sobre o que fazer com os valores atípicos.")
```
